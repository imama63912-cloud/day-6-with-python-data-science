```python
# ==========================================
# LIST AND DICTIONARY COMPREHENSIONS TASK
# Data Science Track - Data Preprocessing
# ==========================================

import numpy as np
import pandas as pd

# Mock datasets matching task suggestions
students_data = {
    "Name": ["Ali", "Sara", "Bilal", "Zahra", "Usman", "Ayesha"],
    "Score": [85, 42, 78, 95, 38, 60],
    "Subject": ["Math", "Math", "Science", "Math", "Science", "Math"],
}

products_data = [
    {"id": 101, "name": "Laptop", "price": 1200, "stock": 15},
    {"id": 102, "name": "Mouse", "price": 25, "stock": 0},
    {"id": 103, "name": "Keyboard", "price": 45, "stock": 8},
    {"id": 104, "name": "Monitor", "price": 300, "stock": 0},
    {"id": 105, "name": "Headphones", "price": 80, "stock": 22},
]


# ------------------------------------------
# SECTION 1: LIST COMPREHENSIONS (5 Examples)
# ------------------------------------------

# Example 1: Basic Transformation (Convert student names to uppercase)
# Traditional Loop Comparison:
# upper_names = []
# for name in students_data["Name"]:
#     upper_names.append(name.upper())
names_uppercase = [name.upper() for name in students_data["Name"]]

# Example 2: Filtering (Extract student scores >= 50 - Passing Marks)
passing_scores = [score for score in students_data["Score"] if score >= 50]

# Example 3: Transformation + Filtering (Scale passing scores by 10%)
scaled_passing_scores = [
    round(score * 1.10, 2) for score in students_data["Score"] if score >= 50
]

# Example 4: Conditional Logic / If-Else (Categorize Pass/Fail status)
grade_status = [
    "Pass" if score >= 50 else "Fail" for score in students_data["Score"]
]

# Example 5: Extraction from List of Dictionaries (In-stock product names)
instock_products = [p["name"] for p in products_data if p["stock"] > 0]


# ----------------------------------------------
# SECTION 2: DICTIONARY COMPREHENSIONS (3 Examples)
# ----------------------------------------------

# Example 1: Mapping Entities (Student Name -> Score mapping)
student_score_map = {
    name: score
    for name, score in zip(students_data["Name"], students_data["Score"])
}

# Example 2: Filtering Dictionary (Product ID -> Price for items > $50)
expensive_products = {
    p["id"]: p["price"] for p in products_data if p["price"] > 50
}

# Example 3: Transformation & Conditional Logic (Categorize stock availability)
inventory_status = {
    p["name"]: ("Available" if p["stock"] > 0 else "Out of Stock")
    for p in products_data
}


# ------------------------------------------
# EXECUTION & RESULTS PRINTING
# ------------------------------------------
if __name__ == "__main__":
    print("--- LIST COMPREHENSIONS ---")
    print("1. Uppercase Names:", names_uppercase)
    print("2. Passing Scores (>= 50):", passing_scores)
    print("3. Scaled Passing Scores (+10%):", scaled_passing_scores)
    print("4. Grade Status (Pass/Fail):", grade_status)
    print("5. In-Stock Products:", instock_products)

    print("\n--- DICTIONARY COMPREHENSIONS ---")
    print("1. Student Score Mapping:", student_score_map)
    print("2. Expensive Products (> $50):", expensive_products)
    print("3. Inventory Status:", inventory_status)
```
